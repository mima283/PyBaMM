```python
# model, ki ima le mehansko degradacijo
model_DFN = pybamm.lithium_ion.DFN(options={"surface form": "differential",
                                                  "particle": "Fickian diffusion",
                                                 "particle mechanics": "swelling only",},
                                        name="Meh.",)


param0 = pybamm.ParameterValues("Ai2020") 
param0.update(
    {"Hydrostatic stress [Pa]": 0},
    check_already_exists=False  
)

param1 = pybamm.ParameterValues("Ai2020") 
param1.update(
    {"Hydrostatic stress [Pa]": 1e3},
    check_already_exists=False  
)

param2 = pybamm.ParameterValues("Ai2020") 
param2.update(
    {"Hydrostatic stress [Pa]": 1e6},
    check_already_exists=False  
)

param3 = pybamm.ParameterValues("Ai2020") 
param3.update(
    {"Hydrostatic stress [Pa]": 1e9},
    check_already_exists=False  
)

param4 = pybamm.ParameterValues("Ai2020") 
param4.update(
    {"Hydrostatic stress [Pa]": 2*1e9},
    check_already_exists=False  
)

param5 = pybamm.ParameterValues("Ai2020") 
param5.update(
    {"Hydrostatic stress [Pa]": 4*1e9},
    check_already_exists=False  
)

param6 = pybamm.ParameterValues("Ai2020")
param6.update(
    {"Hydrostatic stress [Pa]": 6*1e9},
    check_already_exists=False  
)

param7 = pybamm.ParameterValues("Ai2020")
param7.update(
    {"Hydrostatic stress [Pa]": 8*1e9},
    check_already_exists=False  
)

param8 = pybamm.ParameterValues("Ai2020")
param8.update(
    {"Hydrostatic stress [Pa]": 1e10},
    check_already_exists=False  
)

param9 = pybamm.ParameterValues("Ai2020")
param9.update(
    {"Hydrostatic stress [Pa]": 1e11},
    check_already_exists=False  
)
```

```python
var_pts = {
    "x_n": 30,  # negative electrode
    "x_s": 20,  # separator
    "x_p": 30,  # positive electrode
    "r_n": 25,  # negative particle
    "r_p": 25,  # positive particle
}

experiment = pybamm.Experiment(["Discharge at 1 C until 3.0 V"])
solver = pybamm.IDAKLUSolver()  

sim_DFN_0 = pybamm.Simulation(model= model_DFN, parameter_values=param0, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_1 = pybamm.Simulation(model= model_DFN, parameter_values=param1, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_2 = pybamm.Simulation(model= model_DFN, parameter_values=param2, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_3 = pybamm.Simulation(model= model_DFN, parameter_values=param3, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_4 = pybamm.Simulation(model= model_DFN, parameter_values=param4, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_5 = pybamm.Simulation(model= model_DFN, parameter_values=param5, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_6 = pybamm.Simulation(model= model_DFN, parameter_values=param6, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_7 = pybamm.Simulation(model= model_DFN, parameter_values=param7, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_8 = pybamm.Simulation(model= model_DFN, parameter_values=param8, var_pts=var_pts, experiment=experiment, solver=solver)
sim_DFN_9 = pybamm.Simulation(model= model_DFN, parameter_values=param9, var_pts=var_pts, experiment=experiment, solver=solver)



start = time.time()
sol_DFN = sim_DFN_0.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_0 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_1 = sim_DFN_1.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_1 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_2 = sim_DFN_2.solve([0, 3600])  
end = time.time()
print(f"Simulacija DFN_2 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_3 = sim_DFN_3.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_3 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_4 = sim_DFN_4.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_4 zaključena v {end - start:.2} s.")

start = time.time()
sol_DFN_5 = sim_DFN_5.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_5 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_6 = sim_DFN_6.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_6 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_7 = sim_DFN_7.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_7 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_8 = sim_DFN_8.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_8 zaključena v {end - start:.2f} s.")

start = time.time()
sol_DFN_9 = sim_DFN_9.solve([0, 3600])
end = time.time()
print(f"Simulacija DFN_9 zaključena v {end - start:.2f} s.")
```

```python
Q_discharged = sol_DFN["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1] 
DOD = (Q_discharged / Q_max)

Q_discharged = sol_DFN_1["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1] 
DOD_1 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_2["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1] 
DOD_2 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_3["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1]
DOD_3 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_4["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1]
DOD_4 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_5["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1]
DOD_5 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_6["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1]
DOD_6 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_7["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1]
DOD_7 = (Q_discharged / Q_max)

Q_discharged = sol_DFN_8["Discharge capacity [A.h]"].entries
Q_max = Q_discharged[-1]
DOD_8 = (Q_discharged / Q_max)



x_val = [DOD, 
         DOD_1, DOD_2, DOD_3,
          DOD_4, DOD_5, DOD_6, DOD_7,
          DOD_8,
        #   DOD_9
         ]
y_val = [sol_DFN["Voltage [V]"].data, 
         sol_DFN_1["Voltage [V]"].data, sol_DFN_2["Voltage [V]"].data, sol_DFN_3["Voltage [V]"].data,
         sol_DFN_4["Voltage [V]"].data, sol_DFN_5["Voltage [V]"].data, sol_DFN_6["Voltage [V]"].data, 
        sol_DFN_7["Voltage [V]"].data,
          sol_DFN_8["Voltage [V]"].data, 
        #   sol_DFN_9["Voltage [V]"].data
         ]

plt.figure(figsize=fig_size)
for i in range(9):
    plt.plot(x_val[i], y_val[i], label=["$\sigma_{\mathrm{h}}=$0 Pa", 
                                        "$\sigma_{\mathrm{h}}=$1e3 Pa", 
                                        "$\sigma_{\mathrm{h}}=$1e6 Pa", 
                                        "$\sigma_{\mathrm{h}}=$1e9 Pa" , 
                                        "$\sigma_{\mathrm{h}}=$2e9 Pa", 
                                        "$\sigma_{\mathrm{h}}=$4e9 Pa", 
                                        "$\sigma_{\mathrm{h}}=$6e9 Pa",
                                        "$\sigma_{\mathrm{h}}=$8e9 Pa", 
                                        "$\sigma_{\mathrm{h}}=$1e10 Pa", 
                                        # "$\sigma_{\mathrm{h}}=$1e11 Pa"
                                        ][i], 
                                        linestyle=["-", "--", ":", "-", "-", "-", "-", "-", "-", "-"][i])
    
# plt.xlim(3000, 3600)
# plt.ylim(3.5, 3.6)
plt.xlabel("DOD [/]")
plt.ylabel("V [V]")
plt.legend(loc="upper right", bbox_to_anchor=(1.5, 0.9))

if save_fig:
    plt.savefig("grafi/OCV_tlaki.png", format='png', dpi=600, bbox_inches='tight')
```
